**1. line chart**

```python
newdata
```

```python
plt.plot(newdata['batter'],newdata
         ['avg'])

plt.xlabel('Batters name')
plt.ylabel('Batters avg ')
plt.title('Avg. Vs batter')
plt.show()
```

**2. scatter plot**

```python
plt.scatter(newdata['batter'],newdata
         ['avg'])
plt.title('Sactter plot Avg. Vs batter')
```

**3. Bar chart**

```python
plt.bar(newdata['batter'],newdata
         ['avg'])
plt.show()
```

**4. pie chart**

```python
plt.pie(newdata['runs'],labels=newdata
         ['batter'])
plt.title("Runs Distribution (Top 5 Batsmen)")
plt.show()
```
