```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px   # use for to import data set 
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tips.sample(5)
```

```python
# axis level plot
# scatterplot --> relationship between two variables
sns.scatterplot(data=tips, x='total_bill' , y='tip',hue='sex',style='smoker',size='size')
```

```python
# figure level scatter graph 
sns.relplot(data = tips, x='total_bill' , y='tip', kind='scatter',hue='sex',style='smoker',size='size')
```

```python
# line plot    --> trends over time
gap = px.data.gapminder()
```

```python
gap.sample(10)
```

```python
tempdf = gap[gap['country']=='India']
tempdf
```

```python
# axis level plot 
sns.lineplot(data=tempdf,x='year',y='lifeExp')
```

```python
# figure level plot
sns.relplot(data=tempdf,x='year',y='lifeExp',kind='line')
```

```python
tempdf = gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
tempdf
```

```python
sns.relplot(data=tempdf,x='year',y='lifeExp',kind='line',hue='country')
```

```python
# facet plot --> multiple graphs on single col
# facet plot work on relplot[figure level] only
tips
```
