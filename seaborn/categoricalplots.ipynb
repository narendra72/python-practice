<span style="color:gray;">Categorical Plots</span>

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

### Categorical Scatter Plot

- Stripplot
- Swarmplot

```python
# normal scatter plot
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
# 1. strip plot    --> Shows all individual data points  and   Points may overlap
# axes level function
sns.stripplot(data=tips,x='day',y='total_bill')    # 
```

```python
# using catplot   --> figure leve 
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=False,hue='sex')
```

```python
# 2. swarmplot   -->  Similar to strip plot but avoids overlapping and Better visualization of points
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
sns.swarmplot(data=tips, x='day',y='total_bill')
```

### Categorical Distribution Plots

- Boxplot
- Violinplot

```python
# 1. Box plot  --> Shows distribution using quartiles
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
# Using catplot  
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
# single boxplot -> on numerical col
sns.boxplot(data=tips,y='total_bill')
```

```python
# 2. Violinplot = (Boxplot + KDEplot)
# violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

### Categorical Estimate Plot -> for central tendency

- Barplot
- Pointplot
- Countplot

```python
# 1. barplot   --> Shows average (mean) of a numeric variable for each category
sns.barplot(data=tips, x='sex', y='total_bill',hue='smoker')
```

```python
sns.barplot(data=tips, x='sex', y='total_bill',errorbar=None)
```

```python
# 2. point plot  --> Connects points to show trend
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker',errorbar=None)
```

```python
# 3. countplot   --> A general function to create all categorical plots
sns.countplot(data=tips,x='sex',hue='day')
```
