```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#1. histplot
#2. kdeplot
#3. rugplot
```

```python
# for figure --> displot()
tips = sns.load_dataset('tips')
tips
```

```python
# 1. hisplot
sns.histplot(data=tips,x='total_bill',bins=10)
```

```python
sns.histplot(data=tips,x='total_bill',hue='sex')
```

```python
# using displot() at figure kevel
sns.displot(data=tips,x='total_bill',hue='sex',kind='hist')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,kind='hist', x='age')
```

```python
sns.displot(data=titanic,kind='hist', x='age', hue='sex')
```

```python
sns.displot(data=titanic,kind='hist', x='age',col='sex')
```
