```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

<span> Regression Plots </span><br>
<span> 1. regplot() </span><br>
<span> 2. lmplot() </span>

```python
# 1. regplot  --> Shows scatter plot + regression line
sns.regplot(data=tips,x='tip',y='total_bill')
```

```python
# 2. Lmplot  -->Same as regplot() but more advanced
sns.lmplot(data=tips,x='tip',y='total_bill',hue='sex')
```

```python
sns.lmplot(data=tips,x='tip',y='total_bill',hue='sex',ci=None)
```

```python
# residplot() 
sns.residplot(data=tips,x='total_bill', y='tip')
```
